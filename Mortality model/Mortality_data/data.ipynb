```python
data1 = pd.read_csv("/home/kyomukama/ACE/Patient outcomes/Mortality_data/DARING_FINAL_toUse.csv", encoding="iso-8859-1")
```

```python
# Extract antibiotic names from first_ columns
antibiotics = [
    col.replace('first_', '', 1)
    for col in data1.columns
    if col.startswith('first_') and col != 'first_organism'
]

# Select matching first_, sec_, and third_ columns
res_cols = [
    f'{prefix}{antibiotic}'
    for antibiotic in antibiotics
    for prefix in ['first_', 'sec_', 'third_']
    if f'{prefix}{antibiotic}' in data1.columns
]

print(res_cols)
```

```python

# Create Resistance variable
data1['Resistance'] = np.where(
    data1[res_cols].eq('Resistant').any(axis=1),
    'Yes',
    'No'
)
```
